## 1. Build the feature vector

*Code that actually builds it — engineered features, categorical handling, fills.*

```python
# ML 05 — Section 1: Build the feature vector
# Download the starter CSV directly from the public GitHub repository.
# Then load it and build the feature vector.

import pandas as pd
import numpy as np
from pathlib import Path
import urllib.request

# Public GitHub raw file
csv_url = "https://raw.githubusercontent.com/Nayab-khalid/FlyRank-AI-Internship/main/data/raw/content_refresh_anonymized.csv"

local_path = Path("/content/content_refresh_anonymized.csv")

# Download the CSV into the Colab runtime.
if not local_path.exists():
    print("Downloading starter dataset...")
    urllib.request.urlretrieve(csv_url, local_path)

print("Dataset file:", local_path)
print("File exists:", local_path.exists())

# Load the CSV.
df = pd.read_csv(local_path)

print("Dataset loaded successfully.")
print("Rows:", len(df))
print("Columns:", len(df.columns))

# Create the target.
df["is_declining_label"] = (
    df["trend_direction"].astype(str).str.lower() == "down"
).astype(int)

# Numeric features.
numeric_features = [
    "impressions_90d",
    "clicks_90d",
    "sessions_90d",
    "users_90d",
    "engaged_sessions_90d",
    "ai_sessions_90d",
    "days_with_impressions",
    "days_with_sessions",
    "impressions_last_30d",
    "clicks_last_30d",
    "sessions_last_30d",
    "impressions_prev_30d",
    "clicks_prev_30d",
    "sessions_prev_30d",
    "word_count",
    "char_count",
    "content_age_days",
    "days_since_last_update",
    "ctr",
    "avg_position",
    "engagement_rate",
    "scroll_rate",
    "ai_traffic_pct"
]

# Categorical features.
categorical_features = [
    "content_type",
    "main_intent",
    "competition_level"
]

# Make sure every requested feature exists.
missing_features = [
    col for col in numeric_features + categorical_features
    if col not in df.columns
]

if missing_features:
    raise ValueError(
        "These expected columns are missing from the CSV: "
        + str(missing_features)
    )

# Create the feature matrix.
X = df[numeric_features + categorical_features].copy()

# Create the target.
y = df["is_declining_label"].copy()

# Convert numeric columns and fill missing values with the median.
for col in numeric_features:
    X[col] = pd.to_numeric(X[col], errors="coerce")
    X[col] = X[col].fillna(X[col].median())

# Fill categorical missing values.
for col in categorical_features:
    X[col] = X[col].astype("string").fillna("unknown")

print()
print("FEATURE VECTOR CREATED")
print("----------------------")
print("Rows:", len(X))
print("Numeric features:", len(numeric_features))
print("Categorical features:", len(categorical_features))
print("Total features:", len(X.columns))
print("Declining labels:", int(y.sum()))
print("Non declining labels:", int((y == 0).sum()))

display(X.head())
```

## 2. Feature notes (meaning, missing, categorical, available-when?)

*For each feature: what it means, how missing values are handled, and whether it exists BEFORE the moment you predict.*

```python
# ML 05 — Section 2: Feature notes
#
# Feature meaning and availability:
#
# impressions_90d: Search impressions during the 90 day activity window.
# Available from observed activity before the decision.
#
# clicks_90d: Search clicks during the 90 day activity window.
# Available from observed activity before the decision.
#
# sessions_90d: Analytics sessions during the 90 day activity window.
# Available when Analytics data is available.
#
# users_90d: Analytics users during the 90 day activity window.
# Available when Analytics data is available.
#
# engaged_sessions_90d: Engaged Analytics sessions during the 90 day window.
# Available when Analytics data is available.
#
# ai_sessions_90d: Sessions referred from AI tools during the 90 day window.
# Available from observed activity when the relevant Analytics data exists.
#
# days_with_impressions: Number of days with at least one impression.
# Available from the observed activity window.
#
# days_with_sessions: Number of days with at least one session.
# Available from the observed activity window.
#
# impressions_last_30d, clicks_last_30d, sessions_last_30d:
# Activity during the most recent observed 30 day window.
#
# impressions_prev_30d, clicks_prev_30d, sessions_prev_30d:
# Activity during the previous 30 day window.
#
# word_count and char_count:
# Basic content size measures. Missing values are filled with the median.
#
# content_age_days:
# Number of days since the content was created.
#
# days_since_last_update:
# Number of days since the content was last updated.
#
# ctr:
# Click through rate calculated from observed impressions and clicks.
#
# avg_position:
# Average search position during the observed window.
#
# engagement_rate:
# Engagement rate calculated from observed Analytics activity.
#
# scroll_rate:
# Scroll activity relative to page views.
#
# ai_traffic_pct:
# Percentage style measure of AI referred traffic.
#
# content_type:
# Type of content. Missing values are represented as "unknown".
#
# main_intent:
# Search intent category. Missing values are represented as "unknown".
#
# competition_level:
# Keyword competition category. Missing values are represented as "unknown".
#
# Numeric missing values are handled using the median.
# Categorical missing values are handled using the value "unknown".
#
# The feature set is designed around information that is already observed
# at the decision point. Label derived fields are not used as features.
#
# Important limitation:
# The full warehouse has different data availability across clients and
# time periods. Availability flags must be checked when using the warehouse.
#
print("Feature availability review complete.")
print("Numeric missing values: median filled.")
print("Categorical missing values: 'unknown' filled.")
print("Target-derived fields are excluded from X.")
```

## 3. The leakage hunt

*Attack your own features: label-derived columns, future windows, product flags. Show the test.*

```python
# ML 05 — Section 3: Leakage hunt
#
# This cell checks the feature list for fields that must not be used.
#
# The target is derived from trend_direction.
# Therefore trend_direction, trend_pct and is_declining_label are forbidden.
#
# content_id and client_id are identifiers and are not model features.
#
# Future outcome information must also not be included.
#
# Product flags or other downstream outcome fields are also not used.

forbidden_fields = {
    "trend_direction",
    "trend_pct",
    "is_declining_label",
    "content_id",
    "client_id"
}

leaked_fields = sorted(set(X.columns).intersection(forbidden_fields))

print("Forbidden fields found in feature vector:", leaked_fields)
print("Number of forbidden fields:", len(leaked_fields))

if len(leaked_fields) == 0:
    print("LEAKAGE CHECK: PASS")
else:
    print("LEAKAGE CHECK: FAIL")

# Check for names that look like labels, outcomes or future information.
suspicious_keywords = [
    "label",
    "target",
    "future",
    "outcome",
    "trend_direction",
    "trend_pct"
]

suspicious_features = [
    col for col in X.columns
    if any(word in col.lower() for word in suspicious_keywords)
]

print("Suspicious feature names:", suspicious_features)

if len(suspicious_features) == 0:
    print("Feature-name leakage check: PASS")
else:
    print("Feature-name leakage check: REVIEW REQUIRED")

# Privacy check.
privacy_fields = [
    "content_id",
    "client_id"
]

privacy_in_features = [
    col for col in X.columns
    if col in privacy_fields
]

print("Identifier fields used as model features:", privacy_in_features)

if len(privacy_in_features) == 0:
    print("PRIVACY CHECK: PASS")
else:
    print("PRIVACY CHECK: FAIL")
```

## 4. What I excluded and why

*The list of fields you refused to use — with one line of why each.*

```python
# ML 05 — Section 4: What I excluded and why
#
# trend_direction
# Excluded because it directly defines the target.
#
# trend_pct
# Excluded because it is derived from the same trend information used
# to create the target and would therefore leak the outcome.
#
# is_declining_label
# Excluded because it is the target itself.
#
# content_id
# Excluded because it is a pseudonymous identifier, not a useful
# generalizable content signal.
#
# client_id
# Excluded as a predictive feature because it is a pseudonymous identifier.
# It may still be used to create grouped validation splits.
#
# Future outcome fields
# Excluded because information from the future outcome period would make
# the evaluation unrealistic.
#
# Product or downstream decision flags
# Excluded because they may contain information created after the decision
# point and could leak the outcome.
#
# Private client information
# Excluded because this project must remain public safe.
#
# The final feature vector therefore contains measurable content,
# search, engagement, activity and keyword context signals only.

excluded_fields = {
    "trend_direction": "Defines the target",
    "trend_pct": "Derived from the target trend and can leak the outcome",
    "is_declining_label": "The target itself",
    "content_id": "Pseudonymous identifier, not a predictive feature",
    "client_id": "Pseudonymous identifier, used for grouping rather than prediction"
}

for field, reason in excluded_fields.items():
    print(f"{field}: {reason}")

print("\nFinal feature count:", len(X.columns))
print("Feature vector built without label or identifier leakage.")
```

## Correction added later

**This notebook's `LEAKAGE CHECK: PASS` was wrong.**

The check on this page looks for forbidden columns *by name*. It cannot detect a
label that has been rebuilt from columns which are individually allowed, and that
is exactly what my feature vector contained.

`docs/data-dictionary.md` defines the decline label as a drop of more than 20% in
impressions between `impressions_last_30d` and `impressions_prev_30d`. Both columns
are in the 26-feature vector built above, so the label was fully available to the
model while this cell reported PASS.

The check that finds it, and the corrected 20-feature list, are in
`w06_validation_audit.ipynb` Part 3 and `work/leakage_audit.py`. The effect on the
results is reported in [the deployed paper](https://nayab-khalid.github.io/FlyRank-AI-Internship/),
Section 4.

This notebook is left as it ran, because the failure is part of the record.
