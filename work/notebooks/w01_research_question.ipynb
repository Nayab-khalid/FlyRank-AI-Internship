## 1. My lane (or freestyle) and why

```python
# MY LANE: REFRESH / CONTENT OPPORTUNITY SCORING
#
# I chose Refresh / Content Opportunity Scoring because I want to investigate
# which pages should be prioritized for human review based on observable
# search and content signals.
#
# My Week 2 experiments showed that signals such as impressions, content age,
# CTR, and average position can help rank pages for review.
#
# I want to explore whether a leakage-safe approach can improve the
# prioritization of pages for possible refresh, expansion, protection,
# pruning, or monitoring.

import os
import sys
import subprocess

# Download the FlyRank starter repository into Colab
REPO_URL = "https://github.com/flyrank-bih/flyrank-ml-internship-starter"
REPO_DIR = "flyrank-ml-internship-starter"

if not os.path.isdir(REPO_DIR):
    subprocess.run(
        ["git", "clone", "--depth", "1", REPO_URL, REPO_DIR],
        check=True
    )

# Move into the repository
os.chdir(REPO_DIR)

import pandas as pd

# Load the starter dataset
df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

print("Dataset loaded successfully.")
print("Number of pages:", len(df))
```

## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

```python
# THE QUESTION
#
# Question:
# Which pages should a content team review first for refresh, expansion,
# protection, pruning, or monitoring?
#
# Decision:
# The work should help a content team decide which pages deserve limited
# editorial attention first.
#
# Action:
# A content reviewer could investigate the highest-priority pages and decide
# whether to refresh, expand, protect, prune, or continue monitoring them.
#
# Cost of a wrong recommendation:
# A false positive could waste limited editorial time on a page that does not
# need attention. A false negative could cause an important page to be
# overlooked.
#
# Therefore, the output should support human review rather than automatically
# trigger a content change.

print("Decision: prioritize pages for human content review.")
print("Action: review, refresh, expand, protect, prune, or monitor.")
print("Wrong-call cost: wasted editorial effort or missed opportunities.")
```

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
# QUICK LOOK AT THE DATA
#
# I will use actual measurements from the starter dataset to determine whether
# this lane is worth investigating.
#
# The numbers below show:
# 1. Total number of pages
# 2. Proportion of pages currently labelled as declining
# 3. Median impressions over the last 90 days
# 4. Median days since the last update

declining_rate = (
    df["trend_direction"]
    .str.lower()
    .eq("down")
    .mean()
)

print("Number of pages:", len(df))
print("Declining rate:", round(declining_rate, 3))
print("Median impressions (90d):", df["impressions_90d"].median())
print(
    "Median days since last update:",
    df["days_since_last_update"].median()
)
```

## 4. Careful words: what I can and can't claim

*Write what your work will be able to say (observed, directional, decision-support) — and what it never will (causal proof, 'predicting Google').*

```python
# CAREFUL WORDS: WHAT I CAN AND CANNOT CLAIM
#
# I can measure relationships between observable page-level signals and recent
# page performance.
#
# I can test whether a ranking approach prioritizes potentially important
# pages better than a simple baseline.
#
# The output will be directional decision-support, not an automatic decision.
#
# I cannot claim that refreshing a recommended page will cause its performance
# to improve because the data is observational rather than a controlled
# experiment.
#
# I also cannot claim that the model has discovered a Google ranking factor
# or that a page is guaranteed to recover.
#
# Recommendations should therefore be treated as candidates for human review
# and further validation.

print("Can claim: observed, measured, directional, decision-support.")
print("Cannot claim: causation, guaranteed improvement, or prediction of Google's algorithm.")
```
