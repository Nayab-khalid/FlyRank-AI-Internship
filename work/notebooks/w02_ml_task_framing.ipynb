```python
# 1. My lane as an ML task (type)
# My lane is content refresh prioritization.
# I will frame this as a binary classification problem.
# The model will predict whether a content item is showing a declining trend or not.
# Classification fits because each content item has one of two outcomes:
# declining or not declining.
# The prediction supports a real content action by helping the team identify
# which content items should be reviewed for a possible refresh.

print("Task type: Binary classification")
print("Lane: Content refresh prioritization")
```

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

```python
# 2. Target or proxy
# The target is is_declining_label.
# It represents whether a content item is showing a declining trend.
# The label is derived from the observed trend_direction field:
# 1 means trend_direction is "down", and 0 means it is not "down".
# This gives the model a clear binary outcome to predict.
# trend_direction and trend_pct should not be used as model features because
# they are used to define the target and would cause target leakage.

print("Target: is_declining_label")
print("Target: 1 if trend_direction is 'down', otherwise 0")
print("trend_direction and trend_pct should be excluded from model features.")
```

## 3. Success metric

*One metric you can defend. What number means 'good'?*

```python
# 3. Success metric
# The primary success metric is Precision@50.
# It measures the proportion of the top 50 content items selected by the model
# that are actually declining.
# This metric fits the real action because the team has limited time to review
# content, so the quality of the first 50 items is especially important.
# A higher Precision@50 means the top of the review queue contains more
# genuinely declining content items.

k = 50

print("Primary success metric: Precision@50")
print("Top items evaluated:", k)
print("Higher Precision@50 means a better top-50 review queue.")
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

```python
# 5. Why ML beats a fixed rule here
# A fixed rule could identify declining content using a manually chosen threshold.
# However, content performance can depend on several signals at the same time.
# A single threshold may miss useful relationships between these signals.
# ML can learn patterns across multiple available features and use them together
# to predict which content items are more likely to be declining.
# A fixed rule is still useful as a transparent baseline.
# ML is worth testing if it can improve the quality of the content review queue.

print("Why ML?")
print("Content performance depends on multiple signals.")
print("A fixed threshold may miss relationships between those signals.")
print("ML can learn patterns across multiple features.")
print("A fixed rule remains useful as a transparent baseline.")
```
